```python
import json
from pathlib import Path
```

```python
with open(
    "../output/evidence_packet.json",
    "r",
    encoding="utf-8"
) as f:
    evidence = json.load(f)

with open(
    "../output/pader_draft.md",
    "r",
    encoding="utf-8"
) as f:
    pader_draft = f.read()

with open(
    "../output/validation_report.json",
    "r",
    encoding="utf-8"
) as f:
    validation = json.load(f)

print("Evidence loaded")
print("PADER draft loaded")
print("Validation loaded")

print(
    "Validation status:",
    validation["overall_status"]
)
```

```python
if validation["overall_status"] not in ["PASS", "REVIEW"]:
    raise ValueError(
        "Validation failed. "
        "Fix the validation errors before generating "
        "the final PADER."
    )

print("Validation gate passed.")
```

```python
# report info
reporting_period = evidence[
    "reporting_period"
]

case_summary = evidence[
    "case_summary"
]

print("Reporting period:")
print(
    reporting_period["start"],
    "to",
    reporting_period["end"]
)

print("\nCase summary:")
print(
    json.dumps(
        case_summary,
        indent=4
    )
)
```

```python
# report header
final_header = f"""
# Periodic Adverse Drug Experience Report

## Product: Bisoprolol

**Reporting Period:** {reporting_period["start"]} to {reporting_period["end"]}

**Total Unique Cases:** {case_summary["total_cases"]}

**Serious Cases:** {case_summary["serious_cases"]}

**Expedited Cases:** {case_summary["expedited_cases"]}

**Fatal Cases:** {case_summary["fatal_cases"]}

---

"""
```

```python
final_pader = (
    final_header
    + pader_draft
)

print(final_pader)
```

```python
validation_section = f"""

---

## Report Quality Check

The report was generated from the deterministic evidence
packet and subjected to automated validation.

**Validation status:** {validation["overall_status"]}

"""
```

```python
final_pader += validation_section
```

```python
# final markdown report
final_md_path = Path(
    "../output/final_pader.md"
)

with open(
    final_md_path,
    "w",
    encoding="utf-8"
) as f:
    f.write(final_pader)

print(
    "Saved:",
    final_md_path
)
```

```python
final_report = {
    "metadata": {
        "product": "Bisoprolol",
        "reporting_period": reporting_period,
        "validation_status": validation["overall_status"]
    },

    "case_summary": case_summary,

    "demographics": evidence.get(
        "demographics",
        {}
    ),

    "reactions": evidence.get(
        "reactions",
        {}
    ),

    "outcomes": evidence.get(
        "outcomes",
        {}
    ),

    "seriousness_criteria": evidence.get(
        "seriousness_criteria",
        {}
    ),

    "time_trends": evidence.get(
        "time_trends",
        {}
    ),

    "narrative": pader_draft
}
```

```python
# save json
final_json_path = Path(
    "../output/final_pader.json"
)

with open(
    final_json_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        final_report,
        f,
        indent=4,
        ensure_ascii=False
    )

print(
    "Saved:",
    final_json_path
)
```

```python
# final check
print("=" * 70)
print("FINAL PADER GENERATED")
print("=" * 70)

print(
    "Product: Bisoprolol"
)

print(
    "Reporting period:",
    reporting_period["start"],
    "→",
    reporting_period["end"]
)

print(
    "Total cases:",
    case_summary["total_cases"]
)

print(
    "Serious cases:",
    case_summary["serious_cases"]
)

print(
    "Expedited cases:",
    case_summary["expedited_cases"]
)

print(
    "Fatal cases:",
    case_summary["fatal_cases"]
)

print(
    "Validation:",
    validation["overall_status"]
)

print("\nFiles:")
print(final_md_path)
print(final_json_path)
```
