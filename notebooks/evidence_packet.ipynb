```python
import json
import os
from pathlib import Path
```

```python
RESULTS_PATH = "../output/analysis_results.json"

with open(
    RESULTS_PATH,
    "r",
    encoding="utf-8"
) as f:
    analysis = json.load(f)

print("Analysis results loaded.")
print("Sections:")
print(list(analysis.keys()))
```

```python
# creating evidence packet
evidence_packet = {
    "document_type": "Periodic Adverse Drug Experience Report",

    "product": {
        "name": "Bisoprolol"
    },

    "reporting_period": analysis["reporting_period"],

    "dataset": analysis["dataset"],

    "case_summary": {
        "total_cases": analysis["dataset"]["unique_cases"],
        "serious_cases": analysis["seriousness"]["serious_cases"],
        "non_serious_cases": analysis["seriousness"]["non_serious_cases"],
        "expedited_cases": analysis["expedited"]["expedited_cases"],
        "fatal_cases": analysis["fatal_cases"]
    },

    "demographics": {
        "age_distribution": analysis["age_distribution"],
        "age_percentage": analysis["age_percentage"],
        "sex_distribution": analysis["sex_distribution"],
        "sex_percentage": analysis["sex_percentage"],
        "country_distribution": analysis["country_distribution"],
        "country_percentage": analysis["country_percentage"]
    },

    "reactions": {
        "overall": analysis["top_reactions"],
        "serious": analysis["top_serious_reactions"],
        "expedited": analysis["top_expedited_reactions"],
        "fatal_cases": analysis["top_fatal_case_reactions"]
    },

    "outcomes": {
        "counts": analysis["outcomes"],
        "percentages": analysis["outcome_percentage"]
    },

    "seriousness_criteria": analysis[
        "seriousness_criteria"
    ],

    "time_trends": {
        "monthly_cases": analysis["monthly_cases"],
        "monthly_serious_cases": analysis[
            "monthly_serious_cases"
        ]
    }
}

print("Evidence packet created.")
```

```python
# analysis rules
evidence_packet["analysis_rules"] = {
    "source_of_truth": (
        "The deterministic analysis is the sole "
        "source of numerical facts."
    ),

    "no_invention": (
        "Do not invent cases, percentages, reactions, "
        "dates, outcomes, or clinical conclusions."
    ),

    "preserve_uncertainty": (
        "Unknown or missing values must remain unknown."
    ),

    "age_800_code": (
        "The numeric age-unit code 800 was not interpreted "
        "and is therefore included in Unknown age."
    ),

    "country_values": (
        "Country values are preserved as supplied by the dataset."
    ),

    "multiple_outcomes": (
        "Outcome percentages may exceed 100 percent in total "
        "because a case may contain multiple reactions/outcomes."
    ),

    "descriptive_only": (
        "Observed reporting frequencies must not be presented "
        "as evidence of causality."
    )
}
```

```python
OUTPUT_DIR = "../output"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

EVIDENCE_PATH = os.path.join(
    OUTPUT_DIR,
    "evidence_packet.json"
)

with open(
    EVIDENCE_PATH,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        evidence_packet,
        f,
        indent=4,
        ensure_ascii=False
    )

print(
    "Evidence packet saved to:",
    EVIDENCE_PATH
)
```
